```python
import cv2
from djitellopy import tello
import pygame
import time

drone = tello.Tello()
drone.connect()

try:
    print(f"Battery = {drone.get_battery()}")
except BaseException:
    print(f"Battery = 100 %")    
 
# Image Capture

try:
    if drone:
        drone.streamon()

        while True:
            img = drone.get_frame_read().frame()
            image = cv2.resize(img, (300, 300))
            cv2.imshow(winname="Image Capture Drone", mat=image)
            cv2.waitKey(0)
    else:
        print(f"Failed to Capture Image, The Drone doesn't Exist")        
except BaseException as e:
    print(f"Failed to Capture Image, An Error has Occured = {e}")

# Keyboard Control    

def init():
    pygame.init()
    win = pygame.display.set_mode((450,350))

def getkey(keyName):
    ans = False
    for eve in pygame.event.get():pass
    keyInput = pygame.key.get_pressed()
    myKey = getattr(pygame, 'K_{}'.format(keyName))
    if keyInput[myKey]:
        ans = True
    pygame.display.update    
    return ans

def main():
    if getkey("LEFT"):
        print(f"Left Key Pressed")
    elif getkey("RIGHT"):
        print(f"Right Key Pressed")        

def getKeyboardInput():
    lr, fb, ud, yw = 0,0,0,0
    speed = 50

    if getkey("e"): drone.takeoff()

    if getkey("LEFT"): lr = -speed
    elif getkey("RIGHT"): lr = speed

    if getkey("UP"): fb = speed
    elif getkey("DOWN"): fb = -speed

    if getkey("w"): ud = speed
    elif getkey("s"): ud = -speed    

    if getkey("a"): yw = speed
    elif getkey("d"): yw = -speed                         

    if getkey("q"): drone.land()
    return [lr,fb,ud,yw]

init()
try:
    if drone:
        while True:
            values = getKeyboardInput()
            drone.send_rc_control(values[0], values[1], values[2], values[3])
            time.sleep(0.05)
    else:
        print(f"Your Drone doesn't Exist")
except BaseException:
    print(f"An Error has Occured = {e}")            
```
