```python
import djitellopy
import cv2
import time
from djitellopy import tello

drone = tello.Tello()
drone.connect()

try:
    print(f"Battery = {drone.get_battery()}")
except BaseException:
    print(f"Battery = 100 %")    

# Automated Movement

drone.takeoff()
drone.move_forward(50)
drone.move_back(30)
drone.move_left(50)
drone.move_right(20)

# Computer & App Controlled Movement

drone.send_rc_control(2, 4, 3, 1) # Temporary Movement by the Computer
time.sleep(2)
drone.send_rc_control(0, 0, 0, 0) 
time.sleep(2)
drone.send_rc_control(2, 4, 3, 1)
time.sleep(2)
drone.send_rc_control(0, 0, 0, 0)
time.sleep(1)
drone.send_rc_control(2, 4, 3, 1)
time.sleep(2)
drone.send_rc_control(0, 0, 0, 0)
time.sleep(2)
drone.send_rc_control(2, 4, 3, 1)
time.sleep(2)

drone.send_rc_control(0, 0, 0, 0)

try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("Script Closed")
```
